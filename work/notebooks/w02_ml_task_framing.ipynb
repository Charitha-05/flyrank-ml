## 2. Target or proxy

The model will predict a **refresh priority score** for each webpage, representing how strongly it should be prioritized for content review.

Since the FlyRank warehouse dataset does not contain a direct label indicating whether a page "needs refresh," the target will be created as a **defined proxy** using measurable future performance outcomes. The model will learn from historical content characteristics, search performance, engagement metrics, and query-level behavior to estimate which pages are most likely to require attention, helping the SEO team prioritize their review efforts.

## 3. Success metric

The primary success metric for this project is **NDCG@20 (Normalized Discounted Cumulative Gain at 20)**, with **Precision@20** used as a secondary metric.

Since the objective of the project is to **rank webpages by refresh priority**, the order of the recommendations is just as important as identifying the correct pages. NDCG@20 evaluates both the relevance of the recommended pages and whether the most valuable pages are placed near the top of the ranking, where they are most likely to be reviewed by the SEO team.

Precision@20 is used as a secondary metric because it measures how many of the top 20 recommended pages are relevant, providing an intuitive business-oriented measure of recommendation quality. Together, these metrics evaluate both the quality of the ranking and the usefulness of the recommendations for decision-making.

## 5. Why ML beats a fixed rule here

A fixed rule, such as refreshing every page older than a certain number of days or every page with low impressions, cannot capture the complex relationships between multiple factors that influence search performance. The priority of a webpage depends on the combined effect of content characteristics, search visibility, user engagement, keyword behavior, and historical trends. A machine learning model can learn these interactions from data and generate a more reliable priority score than a single manually defined rule. The recommendations are intended to support the SEO team's decisions rather than replace human judgment.
